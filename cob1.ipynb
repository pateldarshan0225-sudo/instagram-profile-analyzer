```python
with open("finaldata.txt","r",encoding="utf-8") as file:
    data = file.read()
```

```python
chunks = data.split("\n\n")
```

```python
chunks = [c for c in chunks if len(c)>3]
```

```python
def parse_chunk(chunk):
    
    chunk = chunk.strip()
    
    sep_chunk = chunk.split('\n')
    
    username = sep_chunk[0]
    
    post = int(sep_chunk[1].split(" post")[0].replace(",", ""))
    
    follower = float(sep_chunk[2].split(" followers")[0].replace(",", "").replace("K","").replace("M",""))
    
    if("K" in sep_chunk[2]):
        follower = int(follower*1000)
    elif("M" in sep_chunk[2]):
        follower = int(follower*1000000)
    else:
        follower = int(follower)
        
    following = float(sep_chunk[3].split(" following")[0].replace(",", "").replace("K","").replace("M",""))

    if("K" in sep_chunk[3]):
        following = int(following*1000)
    elif("M" in sep_chunk[3]):
        following = int(following*1000000)
    else:
        following = int(following)
        
    name = sep_chunk[4]
    
    if(len(sep_chunk)>5):
        account_type = sep_chunk[5]
        bio = sep_chunk[6:]
    else:
        account_type = "Unknown"
        bio = ""
    return {"username": username, "posts": post, "followers": follower,"following": following,"name": name,"account_type":account_type,"bio": bio}
```

```python
all_chunks = []
```

```python
for chunk in chunks:
    parsed_chunk=parse_chunk(chunk)
    all_chunks.append(parsed_chunk)
```
